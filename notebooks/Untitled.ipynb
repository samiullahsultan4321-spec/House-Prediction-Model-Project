```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.read_csv('../data/raw_data/zameen_data.csv')
df = df1.copy()
```

```python
df.head()
df.columns
df.shape
df.isnull().sum()
# only agent and agency has null values........
df.duplicated().sum()
# overall no duplicate record
df.dtypes
```

```python
# Renaming some columns...
df = df.rename(columns={
    'Area Type': 'area_type',
    'Area Size': 'area_size',
    'Area Category': 'area_category'
})
```

```python
# NOW we'll change data types where needed...
df['date_added'] = pd.to_datetime(df['date_added'])
df['property_id'].astype(object)
df['location_id'].astype(object)
df['area'] = np.where(df['area_type'].str.lower()=='kanal', df['area_size']*20,df['area_size'])
df[['area','area_size','area_type']].head(20)
df.drop(columns = ['area_size','area_type'], inplace=True)
```

```python
print("Purpose: ",df['purpose'].unique())
print("Cities: ",df['city'].unique())
print("Purpose Details: \n",df['purpose'].describe())
```

```python
# Now we'll check some columns' range
df['baths'].value_counts().sort_index()
df[df['baths']>11]
df.drop(index=[129605,136139,150484, 158148], inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
df.drop(columns=['agency','agent','page_url','area_category','property_id'],inplace=True)
df.columns
```

```python
df.to_csv('../data/processed_data/cleaned_data.csv',index=False)
```
